```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os
import cv2
```

```python
DATA_PATH = 'Bone Fractures Detection/'

TRAIN_PATH = f'{DATA_PATH}train/'
IMAGE_PATH = f'{TRAIN_PATH}images/'
LABEL_PATH = f'{TRAIN_PATH}labels/'
```

```python
print("Images:", len(os.listdir(IMAGE_PATH)))
print("Labels:", len(os.listdir(LABEL_PATH)))
```

```python
label_files = [f for f in os.listdir(LABEL_PATH) if f.endswith('.txt')]
print('files sum:', len(label_files))
with open(f'{LABEL_PATH}{label_files[0]}', 'r')as f:
    print(f.read())
```

```python
classes = set()
objects_per_image = []
for file in label_files:
    with open (f'{LABEL_PATH}{file}', 'r')as f:
        lines = f.readlines()
        objects_per_image.append(len(lines))
        for line in lines:
            values = line.strip().split()

            if len(values) == 5:
                classes.add(int(values[0]))
```

```python
print("Классы:", sorted(classes))
print("Максимум объектов:", max(objects_per_image))
print("Минимум объектов:", min(objects_per_image))
print("Изображений без объектов:", objects_per_image.count(0))
```

```python
X_train = []
y_train = []

IMG_SIZE = 128

for image_name in os.listdir(IMAGE_PATH):
    try:
        image = cv2.imread(f'{IMAGE_PATH}{image_name}')

        if image is None:
            continue

        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        image = cv2.resize(image, (IMG_SIZE, IMG_SIZE))
        image = image / 255.0

        label_name = os.path.splitext(image_name)[0] + '.txt'

        with open(f'{LABEL_PATH}{label_name}', 'r') as f:
            lines = f.readlines()

        if len(lines) == 0:
            continue

        label = int(lines[0].split()[0])

        X_train.append(image)
        y_train.append(label)

    except:
        continue

X_train = np.array(X_train, dtype=np.float32)
y_train = np.array(y_train)

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
```

```python
from keras.models import Sequential
from keras.layers import (
    Input,
    Conv2D,
    MaxPooling2D,
    BatchNormalization,
    Dropout,
    Flatten,
    Dense
)
from keras.callbacks import EarlyStopping
```

```python
model = Sequential([
    Input(shape=(128, 128, 3)),

    Conv2D(32, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),

    Conv2D(128, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(128, activation='relu'),
    Dropout(0.5),

    Dense(10, activation='softmax')
])

model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
VALID_PATH = f'{DATA_PATH}valid/'

VALID_IMAGE_PATH = f'{VALID_PATH}images/'
VALID_LABEL_PATH = f'{VALID_PATH}labels/'

X_valid = []
y_valid = []

for image_name in os.listdir(VALID_IMAGE_PATH):
    image = cv2.imread(f'{VALID_IMAGE_PATH}{image_name}')

    if image is None:
        continue

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image = cv2.resize(image, (128, 128))
    image = image / 255.0

    label_name = os.path.splitext(image_name)[0] + '.txt'

    with open(f'{VALID_LABEL_PATH}{label_name}', 'r') as f:
        lines = f.readlines()

    if len(lines) == 0:
        continue

    label = int(lines[0].split()[0])

    X_valid.append(image)
    y_valid.append(label)

X_valid = np.array(X_valid, dtype=np.float32)
y_valid = np.array(y_valid)

print(X_valid.shape)
print(y_valid.shape)
```

```python
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    validation_data=(X_valid, y_valid),
    epochs=10,
    batch_size=32,
    callbacks=[early_stop]
)
```

```python
history2 = model.fit(
    X_train,
    y_train,
    validation_data=(X_valid, y_valid),
    epochs=10,
    batch_size=32,
    callbacks=[early_stop]
)
```

```python
history3 = model.fit(
    X_train,
    y_train,
    validation_data=(X_valid, y_valid),
    epochs=10,
    batch_size=32,
    callbacks=[early_stop]
)
```

```python
histories = [history, history2, history3]

accuracy = []
val_accuracy = []
loss = []
val_loss = []

for h in histories:
    accuracy.extend(h.history['accuracy'])
    val_accuracy.extend(h.history['val_accuracy'])
    loss.extend(h.history['loss'])
    val_loss.extend(h.history['val_loss'])
```

```python
epochs = range(1, len(accuracy) + 1)

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(epochs, accuracy, label='Train Accuracy')
plt.plot(epochs, val_accuracy, label='Validation Accuracy')
plt.title('CNN Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(epochs, loss, label='Train Loss')
plt.plot(epochs, val_loss, label='Validation Loss')
plt.title('CNN Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
X_train_binary = []
y_train_binary = []

for image_name in os.listdir(IMAGE_PATH):
    image = cv2.imread(f'{IMAGE_PATH}{image_name}')

    if image is None:
        continue

    label_name = os.path.splitext(image_name)[0] + '.txt'

    with open(f'{LABEL_PATH}{label_name}', 'r') as f:
        lines = f.readlines()

    if len(lines) == 0:
        continue

    classes = [int(line.split()[0]) for line in lines if line.strip()]

    # Есть хотя бы один тип перелома
    label = 1 if any(c != 2 for c in classes) else 0

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image = cv2.resize(image, (128, 128))
    image = image / 255.0

    X_train_binary.append(image)
    y_train_binary.append(label)

X_train_binary = np.array(X_train_binary, dtype=np.float32)
y_train_binary = np.array(y_train_binary)

print("X_train:", X_train_binary.shape)
print("y_train:", y_train_binary.shape)

print("Healthy:", np.sum(y_train_binary == 0))
print("Fractured:", np.sum(y_train_binary == 1))
```

```python
binary_model = Sequential([
    Input(shape=(128, 128, 3)),

    Conv2D(32, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),

    Conv2D(128, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(128, activation='relu'),
    Dropout(0.5),

    Dense(1, activation='sigmoid')
])
```

```python
from keras.optimizers import Adam

binary_model.compile(
    optimizer=Adam(learning_rate=0.0001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

binary_model.summary()
```

```python
def load_binary_data(image_path, label_path):
    X = []
    y = []

    for image_name in os.listdir(image_path):
        image = cv2.imread(f'{image_path}{image_name}')

        if image is None:
            continue

        label_name = os.path.splitext(image_name)[0] + '.txt'
        file_path = f'{label_path}{label_name}'

        if not os.path.isfile(file_path):
            continue

        with open(file_path, 'r') as f:
            lines = [line.strip() for line in f if line.strip()]

        if not lines:
            continue

        classes = [int(line.split()[0]) for line in lines]

        label = 1 if any(c != 2 for c in classes) else 0

        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        image = cv2.resize(image, (128, 128))
        image = image / 255.0

        X.append(image)
        y.append(label)

    return np.array(X, dtype=np.float32), np.array(y)


X_valid_binary, y_valid_binary = load_binary_data(
    f'{DATA_PATH}valid/images/',
    f'{DATA_PATH}valid/labels/'
)

print(X_valid_binary.shape)
print(y_valid_binary.shape)
print("Healthy:", np.sum(y_valid_binary == 0))
print("Fractured:", np.sum(y_valid_binary == 1))
```

```python
early_stop_binary = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

history_binary = binary_model.fit(
    X_train_binary,
    y_train_binary,
    validation_data=(X_valid_binary, y_valid_binary),
    epochs=20,
    batch_size=32,
    callbacks=[early_stop_binary]
)
```

```python
import matplotlib.pyplot as plt

epochs = range(1, len(history_binary.history['accuracy']) + 1)

plt.figure(figsize=(14, 5))

# Accuracy
plt.subplot(1, 2, 1)
plt.plot(epochs, history_binary.history['accuracy'],
         label='Train Accuracy', marker='o')
plt.plot(epochs, history_binary.history['val_accuracy'],
         label='Validation Accuracy', marker='o')

plt.title('Binary CNN - Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

# Loss
plt.subplot(1, 2, 2)
plt.plot(epochs, history_binary.history['loss'],
         label='Train Loss', marker='o')
plt.plot(epochs, history_binary.history['val_loss'],
         label='Validation Loss', marker='o')

plt.title('Binary CNN - Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
X_bbox = []
y_bbox = []

for image_name in os.listdir(IMAGE_PATH):
    image = cv2.imread(f'{IMAGE_PATH}{image_name}')

    if image is None:
        continue

    label_name = os.path.splitext(image_name)[0] + '.txt'
    label_file = f'{LABEL_PATH}{label_name}'

    if not os.path.isfile(label_file):
        continue

    with open(label_file, 'r') as f:
        lines = [line.strip() for line in f if line.strip()]

    # Берём первый объект, который является переломом
    fractures = [
        list(map(float, line.split()))
        for line in lines
        if len(line.split()) == 5 and int(line.split()[0]) != 2
    ]

    if not fractures:
        continue

    bbox = fractures[0][1:5]

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image = cv2.resize(image, (IMG_SIZE, IMG_SIZE))
    image = image / 255.0

    X_bbox.append(image)
    y_bbox.append(bbox)

X_bbox = np.array(X_bbox, dtype=np.float32)
y_bbox = np.array(y_bbox, dtype=np.float32)

print("X:", X_bbox.shape)
print("y:", y_bbox.shape)
```

```python
from keras.models import Sequential
from keras.layers import (
    Input, Conv2D, MaxPooling2D,
    BatchNormalization, GlobalAveragePooling2D,
    Dense, Dropout
)
from keras.optimizers import Adam

bbox_model = Sequential([
    Input(shape=(128, 128, 3)),

    Conv2D(32, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),

    Conv2D(128, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),

    GlobalAveragePooling2D(),

    Dense(128, activation='relu'),
    Dropout(0.3),

    Dense(4, activation='sigmoid')
])

bbox_model.summary()
```

```python
bbox_model.compile(
    optimizer=Adam(learning_rate=0.0001),
    loss='mse',
    metrics=['mae']
)
```

```python
def load_bbox_data(image_path, label_path):
    X = []
    y = []

    for image_name in os.listdir(image_path):
        image = cv2.imread(f'{image_path}{image_name}')

        if image is None:
            continue

        label_name = os.path.splitext(image_name)[0] + '.txt'
        label_file = f'{label_path}{label_name}'

        if not os.path.isfile(label_file):
            continue

        with open(label_file, 'r') as f:
            lines = [line.strip() for line in f if line.strip()]

        fractures = [
            list(map(float, line.split()))
            for line in lines
            if len(line.split()) == 5 and int(line.split()[0]) != 2
        ]

        if not fractures:
            continue

        bbox = fractures[0][1:5]

        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        image = cv2.resize(image, (128, 128))
        image = image / 255.0

        X.append(image)
        y.append(bbox)

    return (
        np.array(X, dtype=np.float32),
        np.array(y, dtype=np.float32)
    )


X_valid_bbox, y_valid_bbox = load_bbox_data(
    f'{DATA_PATH}valid/images/',
    f'{DATA_PATH}valid/labels/'
)

print(X_valid_bbox.shape)
print(y_valid_bbox.shape)
```

```python
from keras.callbacks import EarlyStopping

early_stop_bbox = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

history_bbox = bbox_model.fit(
    X_bbox,
    y_bbox,
    validation_data=(X_valid_bbox, y_valid_bbox),
    epochs=30,
    batch_size=32,
    callbacks=[early_stop_bbox]
)
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

i = 0

prediction = bbox_model.predict(X_valid_bbox[i:i+1], verbose=0)[0]

x, y, w, h = prediction

fig, ax = plt.subplots(figsize=(7, 7))
ax.imshow(X_valid_bbox[i])

x1 = (x - w / 2) * IMG_SIZE
y1 = (y - h / 2) * IMG_SIZE

rect = patches.Rectangle(
    (x1, y1),
    w * IMG_SIZE,
    h * IMG_SIZE,
    linewidth=2,
    edgecolor='red',
    facecolor='none'
)

ax.add_patch(rect)
ax.set_title('Predicted Fracture Location')
ax.axis('off')

plt.show()
```
